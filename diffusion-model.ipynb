```python
import diffusers.models.unets.unet_2d_blocks
import torch
```

```python
layer = diffusers.models.unets.unet_2d_blocks.DownBlock2D(
    in_channels = 4,
    out_channels = 12,
    temb_channels = 128,
    resnet_groups = 2,
    num_layers = 4
)
image = torch.randn((15, 4, 256, 256))
timestep = torch.randn((15, 128))
extremity, trajectory = layer(image, timestep)
extremity.shape
```

```python
layer = diffusers.models.unets.unet_2d_blocks.UNetMidBlock2D(
    in_channels = 8,
    temb_channels = 128,
    resnet_groups = 2,
    num_layers = 1
)
extremity = layer(extremity, timestep)
```

```python
layer = diffusers.models.unets.unet_2d_blocks.UpBlock2D(
    in_channels=12, prev_output_channel=12, out_channels=6, temb_channels=128, num_layers=1, resnet_groups=2
)
extremity = layer.forward(extremity, trajectory, timestep)
extremity.shape
```

```python

# 定義 DownBlock 和 UpBlock
down_block = diffusers.models.unets.unet_2d_blocks.DownBlock2D(
    in_channels=32,  # 輸入圖像通道數
    out_channels=64,  # 下采樣後的輸出通道數
    temb_channels=128,
    num_layers=1
)
up_block = diffusers.models.unets.unet_2d_blocks.UpBlock2D(
    in_channels=64,  # 跳躍連接特徵圖的通道數
    prev_output_channel=32,  # 下采樣後輸出的特徵圖通道數
    out_channels=256,  # 上采樣輸出的通道數
    temb_channels=128,
    num_layers=2
)

# 模擬輸入
x = torch.randn(1, 32, 64, 64)
t = torch.randn(1, 128)
# 下采樣
x_down, skip_features = down_block(x, t)
print(f"下采樣特徵形狀: {x_down.shape}")  # (1, 64, 32, 32)
print(f"跳躍連接特徵形狀: {skip_features[0].shape}")  # (1, 4, 64, 64)

# 上采樣
x_up = up_block(x_down, skip_features, t)
print(f"上采樣特徵形狀: {x_up.shape}")  # (1, 4, 64, 64)
```
